```python
def get_season(month):
    season_dict = {
        'December': 'Winter', 'January': 'Winter', 'February': 'Winter',
        'March': 'Spring', 'April': 'Spring', 'May': 'Spring',
        'June': 'Summer', 'July': 'Summer', 'August': 'Summer',
        'September': 'Autumn', 'October': 'Autumn', 'November': 'Autumn'
    }
    return season_dict.get(month, 'Unknown')

x_train['fe_season'] = x_train['arrival_date_month'].apply(get_season)
x_test['fe_season'] = x_test['arrival_date_month'].apply(get_season)
```

# Encoding

```python
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler 

def encode_train(df):
    df = df.copy()
    encoders = {}

    for col in df.select_dtypes(exclude=np.number).columns:
        le = LabelEncoder()
        df[col] = le.fit_transform(df[col])
        encoders[col] = le
    return df, encoders

def encode_test(df, encoders, train_columns):
    df = df.copy()

    # Label encoding
    for col, le in encoders.items():
        df[col] = df[col].astype(str)
        df[col] = df[col].apply(lambda x: le.transform([x])[0] if x in le.classes_ else -1)

    # Align columns
    df = df.reindex(columns=train_columns, fill_value=0)

    return df
    
def scale_train(df):
    df = df.copy()
    scalers = {}

    for col in df.select_dtypes(include=np.number).columns:
        scaler = StandardScaler()
        df[col] = scaler.fit_transform(df[[col]])
        scalers[col] = scaler

    return df, scalers

def scale_test(df, scalers):
    df = df.copy()

    for col in df.columns:
        if col in scalers:
            df[col] = scalers[col].transform(df[[col]])

    return df
```

```python
# Encoding
x_train_encoded, encoders = encode_train(X_train_transformed)
x_test_encoded = encode_test(X_test_transformed, encoders, X_train_transformed.columns)
```

```python
# Scaling
x_train_scaled, scalers = scale_train(df=x_train_encoded)
x_test_scaled = scale_test(df=x_test_encoded, scalers=scalers)
```

# Logistic Regression

```python
import pandas as pd

from imblearn.pipeline import Pipeline
from imblearn.over_sampling import SMOTE

from sklearn.linear_model import LogisticRegression
from sklearn.feature_selection import SelectFromModel

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)
```

```python
selector = SelectFromModel(
    LogisticRegression(
        penalty="l1",
        solver="liblinear",
        C=1.0,
        max_iter=1000,
        random_state=42
    ),
    threshold="mean"
)

pipeline = Pipeline([
    ("selector", selector),

    ("sampler", SMOTE(
        random_state=42
    )),

    ("model", LogisticRegression(
        penalty="l1",
        solver="liblinear",
        C=1.0,
        max_iter=1000,
        random_state=42
    ))
])
```

```python
pipeline
```

```python
pipeline.fit(
    x_train_scaled,
    y_train
)
```

```python
# Before Selector 
X_train_selected = pipeline.named_steps["selector"].transform(
    x_train_scaled
)

# After SMOTE
X_train_smote, y_train_smote = pipeline.named_steps["sampler"].fit_resample(
    X_train_selected,
    y_train
)

print("Before SMOTE:")
print(y_train.value_counts())

print("\nAfter SMOTE:")
print(pd.Series(y_train_smote).value_counts())

print("\nShape before SMOTE:")
print(X_train_selected.shape)

print("\nShape after SMOTE:")
print(X_train_smote.shape)
```

```python
selected_mask = pipeline.named_steps["selector"].get_support()
selected_features = x_train_scaled.columns[selected_mask]

print(f"Number of selected features: {len(selected_features)} / {x_train_scaled.shape[1]}")
print(selected_features.tolist())
```

```python
dropped_features = x_train_scaled.columns[~selected_mask]
print(f"Dropped features: {dropped_features.tolist()}")
```

```python
coefs = pipeline.named_steps["selector"].estimator_.coef_[0]
feature_importance = pd.Series(coefs, index=x_train_scaled.columns).sort_values(key=abs, ascending=False)
print(feature_importance)
```

```python
y_pred = pipeline.predict(x_test_scaled)

y_proba = pipeline.predict_proba(
    x_test_scaled
)[:, 1] 
```
