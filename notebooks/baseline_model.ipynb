# Hotel Booking Cancellation - Baseline Model

## Purpose
During the EDA stage, we analyzed the data and identified key factors influencing 
cancellations (e.g., `customer_type`, `meal`, `lead_time`, `deposit_type`, etc.). 
Now, based on these insights, we need to build a **baseline model**.

## Why do we need a baseline model?
A baseline model is the simplest, fastest model to build, and it serves the 
following purposes:

1. **Establish a benchmark** — to compare more complex models (e.g., XGBoost, 
   Random Forest) against this baseline and confirm whether they genuinely 
   perform better.
2. **Validate the pipeline** — to ensure data preprocessing, train/test split, 
   and evaluation logic work correctly before moving to more complex modeling.
3. **Get a quick first result** — a simple model (e.g., Logistic Regression) 
   trains fast and gives an initial sense of how predictable the target is 
   with the current features.

```python
import pandas as pd
import numpy as np
import joblib
import os

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import MinMaxScaler, OrdinalEncoder

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split

from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
df=pd.read_csv("../data/raw/hotel_bookings_updated_2024.csv")

target="is_canceled"
drop_cols=["company", "reservation_status"]

x=df.drop(columns=[target] + drop_cols)
y=df[target]
```

```python
# Train test split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42 )
```

```python
# Splitting numeric and categorical features
numeric_features=x_train.select_dtypes(include=[np.number]).columns
categorical_features=x_train.select_dtypes(exclude=[np.number]).columns
```

```python
# Pipeline

numeric_pipeline=Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", MinMaxScaler())
])

catgorical_pipeline=Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ('encoder', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1))
])

preprocessor=ColumnTransformer([
    ("num", numeric_pipeline, numeric_features),
    ("cat", catgorical_pipeline, categorical_features)
])

```

```python
# Training

lg_pipeline=Pipeline([
    ("preprocessing", preprocessor),
    ("model", LogisticRegression())
])

dt_pipeline=Pipeline([
    ("preprocessing", preprocessor),
    ("model", DecisionTreeClassifier(
        max_depth=5,
        random_state=42
    ))
])

lg_pipeline.fit(x_train, y_train)
dt_pipeline.fit(x_train, y_train)
```

```python
# Predicts

lg_predicts=lg_pipeline.predict(x_test)
dt_predicts=dt_pipeline.predict(x_test)

# Evaluation

lg_acc_score=accuracy_score(y_test, lg_predicts)
dt_acc_score=accuracy_score(y_test, dt_predicts)
print("Logistic Regression: ", lg_acc_score)
print("Decision Tree Classifier: ", dt_acc_score)
```

```python
lg_predicts_=lg_pipeline.predict(x_train)
dt_predicts_=dt_pipeline.predict(x_train)

lg_acc_score=accuracy_score(y_train, lg_predicts_)
dt_acc_score=accuracy_score(y_train, dt_predicts_)
print("Logistic Regression: ", lg_acc_score)
print("Decision Tree Classifier: ", dt_acc_score)
```

```python
results = pd.concat([x_test[["meal", "market_segment", "reserved_room_type"]], y_test], axis=1)
results["lg_pred"]=lg_predicts
results["dt_pred"]=dt_predicts

results.sort_values("is_canceled", ascending=False).tail(10)
```

```python
# Model saving
from pathlib import Path

ROOT_DIR = Path.cwd().parent

model_dir = ROOT_DIR / "models" / "baseline"
model_dir.mkdir(parents=True, exist_ok=True)
model_file = model_dir / "baseline_model.joblib"

joblib.dump(dt_pipeline, model_file)
```
